Deep and Shallow Copy

```python
a = np.arange(1,11)
a
```

```python
# Deep copy Eg.
a_copy = a.copy()
a_copy[0] = 99
print(a)
print(a_copy)
```

```python
# Shallow copy Eg.
b = a
b[0] = 99
print(a)
print(b)
```

Advance Array Manipulation 

1. Stacking Array


```python
# must be performed on Vectors
# can't be performed on matrix
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
#  Type 1:- Vertical Stack
np.vstack((a,b))
```

```python
# Type 2:- Horizontal Stack
np.hstack((a,b))
```

```python
# Type 3:- Column Stack
np.column_stack((a,b))
```

EXERCISE


```python
# Columns: [Age, Math Marks, Science MArks]

data = np.array([
    [18,85,78],  # Student 1 
    [19,92,88],  # Student 2
    [17,76,95],  # Student 3
    [18,65,70],  # Student 4
    [20,90,85]   # Student 5
])
```

```python
#  Get the shape of the Matrix
data.shape
```

```python
# Find the average age of students
avg = np.mean(data[:,0])
print("Average Age of Students: ",avg)
```

```python
# Extract Math marks of all Students
math_marks = data[:,1]
print("Math Marks:" ,math_marks)
```

```python
#  Find the highest Science mark
sci_hmark = np.max(data[:,2])
print("Highest Science Mrak: ",sci_hmark)
```

```python
# Get details of Student who scored more than 90 in math
data[math_marks > 90]
```

```python
# Increase Math marks of all students by 5
print("Math Marks Increades by 5: ")
data[:,1] +=5
data
```

```python
# Find how many students younger than 18
len(data[data[:,0] < 19])
```

```python
# Calculate the average marks in each subject (column-wise mean)
total = np.mean(data[:,1:], axis=0)
total
```

```python
# Get data of students who scored at least 80 in both subjects
print("Students whos scored at least 80 in both Subjects:")
print(data[(data[:,1] >= 80) & (data[:,2] >= 80) ])
```

```python
# Replace all Science marks < 75 with 0
data[:,2][data[:,2] < 75] = 0
data
```
