```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Matplotlib Basics


```python
x = np.linspace(0,5,11)
y = x **2
```

```python
a = np.linspace(0,10,100)
```

```python
plt.plot(a,np.sin(a))
plt.show()
```

```python
plt.plot(a, np.sin(a))
plt.plot(a, np.cos(a))
plt.show()
```

```python
plt.plot(np.sin(a), color = "g")
plt.show()
```

```python
plt.plot(np.sin(a), color= "0.8")
```

```python
plt.plot(np.sin(a), linestyle="solid")
plt.plot(np.sin(a+1), linestyle="dashed")
plt.plot(np.sin(a+2), linestyle="dashdot")
plt.plot(np.sin(a+3), linestyle="dotted")
plt.show()
```

```python
plt.plot(np.sin(a), "-g")
plt.plot(np.sin(a+1), "--c")
plt.plot(np.sin(a+2), "-.k")
plt.plot(np.sin(a+3), ":r")
plt.show()
```

```python
# Advanced line chart - Multiple lines and legends
y1 = x **2
y2 = x**3

plt.plot(x, y1, label= "Line 1")
plt.plot(x, y2, label= "Line 2")

plt.title("Multi-line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.legend()
plt.show()
```

# Subplots in Matplotlib 

```python
plt.subplot(2,2,1)
plt.plot(y**2, y**2)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2, x**2)
plt.show()
```

```python
# Data 1 for Bar chart 
a = ["Mon","Tue","Wed","Thu","Fri"]
b = [19, 27, 58, 40, 36]

# Data 2 for scatter plot
a1 = ["A","B","C","D","E"]
b1 = [19, 27, 58, 40, 36]

plt.figure(figsize=(20,14))
plt.subplot(2,2,1)
plt.bar(a,b)
plt.title("Bar Chart")
plt.xlabel("Week Days")
plt.ylabel("Students Present")

plt.subplot(2,2,2)
plt.scatter(b,b1)
plt.title("Scatter Plot")
plt.xlabel("Week Days")
plt.ylabel("Students Present")

plt.subplot(2,2,3)
plt.pie(b, labels=a, autopct="%.2f%%")
plt.title("Pie Chart")

plt.subplot(2,2,4)
plt.plot(a,b)
plt.title("Line Graph")
plt.xlabel("Week Days")
plt.ylabel("Students Present")
plt.show()

```

# Object Oriented way

```python
fig = plt.figure()
axis1 = fig.add_axes([0.,0.1,1,1])
axis1.plot(x,y)
plt.show()
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.1,0.1,.8,.8])
axis2.plot(x,y)
plt.show()
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.1,.3,.3])
axis2.plot(x,y)
plt.show()
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,.3,.3])
axis2.plot(x,y)
plt.show()
```

```python
fig = plt.figure(figsize=(20,14),dpi=300)
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,.3,.3])
axis2.plot(x,y)
plt.show()
```

# Performing operations on real data

```python
import pandas as pd

# Sample Monthly Sales Data
sales_data = {
    "Month": [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June",
        "July",
        "August",
        "September",
        "October",
        "November",
        "December",
    ],
    "Sales_USD": [
        12500,
        15000,
        14200,
        18900,
        21000,
        19500,
        23000,
        22100,
        20400,
        25000,
        28500,
        34000,
    ],
    "Units_Sold": [250, 290, 275, 340, 390, 360, 410, 400, 380, 450, 510, 620],
}

# Create DataFrame
df = pd.DataFrame(sales_data)
```

```python
df
```

```python
fig = plt.figure(figsize=(14,10))
plt.plot(df['Month'],df['Sales_USD'])
plt.title("Sales per month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
fig = plt.figure(figsize=(14,10))
plt.bar(df['Month'],df['Sales_USD'])
plt.title("Sales per month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
plt.pie(df['Sales_USD'],labels=df['Month'], autopct="%1.1f%%")
plt.title("Sales Per Month")

plt.show()
```

```python
fig = plt.figure(figsize=(14,10), dpi=300)
plt.stem(df['Month'],df['Sales_USD'])
plt.title("Sales per month")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```
