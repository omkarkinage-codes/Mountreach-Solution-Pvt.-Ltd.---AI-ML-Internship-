## Special Assignment Kaggle Dataset Analysis & Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set global style
sns.set_theme(style="whitegrid")
```

```python

# Load the dataset
df = pd.read_csv("insurance.csv")
```

```python
# Display the first few records
print("First 5 Records ")
display(df.head())
```

```python
# Check dataset shape
print(f"Dataset Shape (Rows, Columns): {df.shape}")
```

```python

# Display column names
print(f"Column Names: {df.columns.tolist()}")
```

```python
# Check data types
print("Data Types ")
print(df.dtypes)
```

```python
# Identify missing values
print("Missing Values Count ")
print(df.isnull().sum())
```

```python
# Generate summary statistics
print("Summary Statistics ")
display(df.describe(include='all'))
```

 1. Histogram: Visualizing the distribution of the target variable (Medical Charges)


```python
plt.figure(figsize=(8, 5))
sns.histplot(df['charges'], kde=True, color='teal')
plt.title('Distribution of Medical Charges')
plt.xlabel('Charges ($)')
plt.ylabel('Count')
plt.show()
```

2. Bar Chart: Comparing the average medical charges between Smokers and Non-Smokers

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x='smoker', y='charges', palette='muted', errorbar=None)
plt.title('Average Charges: Smokers vs Non-Smokers')
plt.xlabel('Smoker Status')
plt.ylabel('Mean Charges ($)')
plt.show()
```

3. Scatter Plot: Analyzing the relationship between Age and Charges, color-coded by Smoking Status

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='age', y='charges', hue='smoker', palette='Set1', alpha=0.7)
plt.title('Age vs Medical Charges')
plt.xlabel('Age')
plt.ylabel('Charges ($)')
plt.show()
```

4. Box Plot: Checking the variance and distribution of Body Mass Index (BMI) across different regions

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='region', y='bmi', palette='Pastel1')
plt.title('BMI Distribution Across Regions')
plt.xlabel('Region')
plt.ylabel('BMI')
plt.show()
```

5. Heatmap / Correlation Matrix: Visualizing the linear correlation between all numerical variables

```python
plt.figure(figsize=(10, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix of Numerical Variables')
plt.show()
```
