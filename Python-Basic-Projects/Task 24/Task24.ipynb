Q1. (Dataset Selection & Preprocessing) 

Choose one classification dataset and one regression dataset (you can use previous datasets or find new ones). 

Load and clean both datasets. 

Perform necessary preprocessing (encoding, scaling, 
train-test split). 

Clearly identify Independent Features (X) and 

Dependent Feature (y) for both. 


```python
import joblib
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_california_housing, fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
diabetes_data = fetch_openml(name="diabetes", version=1, as_frame=True)
X_clf = diabetes_data.data
# Convert categorical target ('tested_positive' / 'tested_negative') to 1 / 0
y_clf = (diabetes_data.target == "tested_positive").astype(int)

# Train-Test Split & Scaling for Classification
X_clf_train, X_clf_test, y_clf_train, y_clf_test = train_test_split(
    X_clf, y_clf, test_size=0.2, random_state=42
)

scaler_clf = StandardScaler()
X_clf_train_scaled = scaler_clf.fit_transform(X_clf_train)
X_clf_test_scaled = scaler_clf.transform(X_clf_test)
```

```python
housing = fetch_california_housing(as_frame=True)
X_reg = housing.data  # Independent Features (X)
y_reg = housing.target  # Dependent Target (y): Median House Value

# Train-Test Split & Scaling for Regression
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)
```

```python
scaler_reg = StandardScaler()
X_reg_train_scaled = scaler_reg.fit_transform(X_reg_train)
X_reg_test_scaled = scaler_reg.transform(X_reg_test)

print("Preprocessing Complete!")
print(f"Classification Data shape: X = {X_clf.shape}, y = {y_clf.shape}")
print(f"Regression Data shape:     X = {X_reg.shape}, y = {y_reg.shape}")
```

Q2. (Classification Algorithms) 

On the classification dataset, train and evaluate the following models: 
•	Logistic Regression 
•	Decision Tree Classifier 
•	Support Vector Machine (SVM) 
•	K-Nearest Neighbors (KNN) 
•	Naive Bayes 

For each model: 
•	Train the model 
•	Make predictions 
•	Calculate Accuracy, Confusion Matrix, and Classification Report 

Create a comparison table of all models’ performance. 





```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
```

```python
clf_models = {
    "Logistic Regression": LogisticRegression(random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "SVM": SVC(probability=True, random_state=42),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Naive Bayes": GaussianNB(),
}

clf_results = []
trained_clf_models = {}
```

```python
# Train and Evaluate each classifier
for name, model in clf_models.items():
    if name == "Decision Tree":
        model.fit(X_clf_train, y_clf_train)
        preds = model.predict(X_clf_test)
    else:
        model.fit(X_clf_train_scaled, y_clf_train)
        preds = model.predict(X_clf_test_scaled)

    trained_clf_models[name] = model

    # Calculate metrics
    acc = accuracy_score(y_clf_test, preds)
    prec = precision_score(y_clf_test, preds)
    rec = recall_score(y_clf_test, preds)
    f1 = f1_score(y_clf_test, preds)

    clf_results.append(
        {
            "Model": name,
            "Accuracy": round(acc, 4),
            "Precision": round(prec, 4),
            "Recall": round(rec, 4),
            "F1 Score": round(f1, 4),
        }
    )

    print(f"{name} ")
    print("Confusion Matrix:\n", confusion_matrix(y_clf_test, preds))
    print(
        "\nClassification Report:\n",
        classification_report(
            y_clf_test, preds, target_names=["Negative", "Positive"]
        ),
    )

```

```python
# Display Comparison Table
df_clf_comparison = pd.DataFrame(clf_results)
print("\nCLASSIFICATION MODEL COMPARISON TABLE ")
df_clf_comparison
```

Q3. (Regression Algorithms) 

On the regression dataset, train and evaluate the following models: 
•	Linear Regression 
•	Decision Tree Regressor 
•	Support Vector Regressor (SVR) 
•	K-Nearest Neighbors Regressor 

For each model: 
•	Train the model 
•	Make predictions 
•	Calculate R² Score 
Create a comparison table showing which model performed best.


```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
```

```python
reg_models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree Regressor": DecisionTreeRegressor(random_state=42),
    "SVR": SVR(),
    "KNN Regressor": KNeighborsRegressor(n_neighbors=5),
}

reg_results = []
trained_reg_models = {}
```

```python
# Train and Evaluate each regressor
for name, model in reg_models.items():
    if name == "Decision Tree Regressor":
        model.fit(X_reg_train, y_reg_train)
        preds = model.predict(X_reg_test)
    else:
        model.fit(X_reg_train_scaled, y_reg_train)
        preds = model.predict(X_reg_test_scaled)

    trained_reg_models[name] = model

    r2 = r2_score(y_reg_test, preds)
    reg_results.append({"Model": name, "R² Score": round(r2, 4)})

# Display Comparison Table sorted by R² Score
df_reg_comparison = pd.DataFrame(reg_results).sort_values(
    by="R² Score", ascending=False
)
```

```python
print("\nREGRESSION MODEL COMPARISON TABLE ")
df_reg_comparison
```

 Q4. (Best Model Selection & Saving) 

From Q2 and Q3: 

Select the best classification model and best regression model based on 
performance metrics. 

Save both best models using joblib (also save scaler and columns if used). 


```python
# Identify and Save Best Classification Model
best_clf_name = df_clf_comparison.sort_values(
    by=["Accuracy", "F1 Score"], ascending=False
).iloc[0]["Model"]
best_clf_model = trained_clf_models[best_clf_name]

joblib.dump(best_clf_model, "best_classification_model.pkl")
joblib.dump(scaler_clf, "scaler_clf.pkl")
joblib.dump(list(X_clf.columns), "columns_clf.pkl")

print(
    f"Best Classification Model: '{best_clf_name}' saved to 'best_classification_model.pkl'"
)
```

```python
# Identify and Save Best Regression Model
best_reg_name = df_reg_comparison.iloc[0]["Model"]
best_reg_model = trained_reg_models[best_reg_name]

joblib.dump(best_reg_model, "best_regression_model.pkl")
joblib.dump(scaler_reg, "scaler_reg.pkl")
joblib.dump(list(X_reg.columns), "columns_reg.pkl")

print(
    f"Best Regression Model: '{best_reg_name}' saved to 'best_regression_model.pkl'"
)
```
