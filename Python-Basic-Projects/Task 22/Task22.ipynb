Q1. (Data Loading & Preprocessing)

Load the Heart Disease dataset.

Separate features into X (independent) and y (dependent/target = HeartDisease).

Perform necessary preprocessing (encoding categorical columns).


```python
import pandas as pd
```

```python
df = pd.read_csv("heart.csv")
```

```python
df.head()
```

```python
# Separate features into X and y
x = df.drop(columns=["HeartDisease"])
y = df["HeartDisease"]
```

```python
# Identify categorical columns
categorical_clos = x.select_dtypes(include=["object"]).columns.tolist()
categorical_clos
```

```python
# Perform One-Hot Encoding
x_encoded = pd.get_dummies(x, columns=categorical_clos, drop_first=True).astype(int)
```

```python
x_encoded.head()
```

Q2. (Train-Test Split)

Split the preprocessed data into training and testing sets using train_test_split.

Use test_size=0.20 and random_state=42.

Print the shape of X_train, X_test, y_train, and y_test.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    x_encoded, y, test_size=0.20, random_state=42
)
```

```python
# Print the shapes of all split subsets
print(f"X_train shape (Training features): {X_train.shape}")
print(f"X_test shape  (Testing features):  {X_test.shape}")
print(f"y_train shape (Training target):   {y_train.shape}")
print(f"y_test shape  (Testing target):    {y_test.shape}")
```

Q3. (Building Logistic Regression Model)

Train a Logistic Regression model:

Import LogisticRegression from sklearn.

Create the model and fit it on the training data (X_train, y_train).


```python
from sklearn.linear_model import LogisticRegression
log_reg = LogisticRegression(max_iter= 1000, random_state= 42)
```

```python
log_reg.fit(X_train, y_train)
```

Q4. (Making Predictions)

Use the trained model to predict on the test data.

Store predictions in y_pred.

Print the first 10 actual values (y_test) and predicted values (y_pred).


```python
# Use the trained model to predict target labels on the test dataset
y_pred = log_reg.predict(X_test)
```

```python
comparison_df = pd.DataFrame(
    {
        "Actual Value (y_test)": y_test.values[:10],
        "Predicted Value (y_pred)": y_pred[:10],
    }
)

# 3. Print the first 10 actual and predicted values
print("First 10 Actual vs Predicted Values ")
print(comparison_df)
```

Q5. (Confusion Matrix)

Create a Confusion Matrix using confusion_matrix(y_test, y_pred).

Display the matrix.

Label TN, FP, FN, and TP in the output.


```python
from sklearn.metrics import confusion_matrix 

cm = confusion_matrix(y_test, y_pred)

tn, fp, fn, tp = cm.ravel()

```

```python
print("Confusion Matrix ")
print(f"[[TN={tn}  FP={fp}]")
print(f" [FN={fn}  TP={tp}]]")
```

Q6. (Model Evaluation Metrics)

Calculate the following metrics:
1. Accuracy
2. Precision
3. Recall
4. F1 Score

Use classification_report(y_test, y_pred) and print the full report.


```python
from sklearn.metrics import(
    accuracy_score,
    classification_report,
    f1_score,
    precision_score,
    recall_score
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
```

```python
print("Individual Performance Metrics ")
print(f"Accuracy  : {accuracy:.4f} ({accuracy * 100:.2f}%)")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
```

```python
print("Full Classification Report ")
report = classification_report(
    y_test, y_pred, target_names=["No Disease (0)", "Heart Disease (1)"]
)
print(report)
```

Q8. (Loading & Testing Saved Model)

Load the saved model and objects.

Create sample input data (one patient record).

Preprocess the input (encoding + scaling).

Make a prediction using the loaded model and print the result.


```python
loaded_model = joblib.load("heart_model.pkl")
loaded_columns = joblib.load("columns.pkl")
```

```python
sample_patient = pd.DataFrame(
    [
        {
            "Age": 54,
            "Sex": "M",
            "ChestPainType": "ATA",
            "RestingBP": 130,
            "Cholesterol": 236,
            "FastingBS": 0,
            "RestingECG": "Normal",
            "MaxHR": 174,
            "ExerciseAngina": "N",
            "Oldpeak": 0.0,
            "ST_Slope": "Up",
        }
    ]
)

print("Unprocessed Sample Patient Record ")
print(sample_patient)
```

```python
categorical_clos = sample_patient.select_dtypes(include=["object"]).columns.tolist()
```

```python
sample_encoded = pd.get_dummies(sample_patient, columns=categorical_clos, drop_first=True).astype(int)
```

```python
sample_aligned = sample_encoded.reindex(columns=loaded_columns, fill_value=0)
```

```python
prediction = loaded_model.predict(sample_aligned)[0]
prediction_proba = loaded_model.predict_proba(sample_aligned)[0]
```

```python
print("Model Prediction Result")
if prediction == 1:
    print(
        f"Target Predicted: 1 (Heart Disease Detected) with {prediction_proba[1] * 100:.2f}% probability."
    )
else:
    print(
        f"Target Predicted: 0 (Normal / No Heart Disease) with {prediction_proba[0] * 100:.2f}% probability."
    )
```

Q9. (Streamlit Input Interface)

Create a Streamlit app (app.py) with:

Input fields for all features using st.number_input(), st.selectbox(), etc.

A "Predict" button.


```python
# app.py
import joblib
import pandas as pd
import streamlit as st

# 1. Streamlit Page Configuration
st.set_page_config(
    page_title="Heart Disease Predictor", page_icon="❤️", layout="centered"
)

# 2. Title and Description
st.title("❤️ Heart Disease Prediction App")
st.write(
    "Fill in the patient's clinical metrics below to predict the likelihood of heart disease."
)
st.markdown("---")

# 3. Load Saved Model Artifacts
try:
    model = joblib.load("heart_model.pkl")
    encoded_columns = joblib.load("columns.pkl")
    data_loaded = True
except FileNotFoundError:
    st.error(
        "❌ Missing required files! Please ensure 'heart_model.pkl' and 'columns.pkl' are in the same folder as this app."
    )
    data_loaded = False

if data_loaded:
    # 4. Input Fields Layout
    st.subheader("Patient Clinical Data")

    col1, col2 = st.columns(2)

    with col1:
        age = st.number_input(
            "Age (Years)", min_value=1, max_value=120, value=50, step=1
        )
        sex = st.selectbox(
            "Gender (Sex)", options=["M", "F"], help="M: Male, F: Female"
        )
        chest_pain = st.selectbox(
            "Chest Pain Type",
            options=["TA", "ATA", "NAP", "ASY"],
            help="TA: Typical Angina, ATA: Atypical Angina, NAP: Non-Anginal Pain, ASY: Asymptomatic",
        )
        resting_bp = st.number_input(
            "Resting Blood Pressure (mm Hg)",
            min_value=50,
            max_value=250,
            value=120,
            step=1,
        )
        cholesterol = st.number_input(
            "Serum Cholesterol (mm/dl)",
            min_value=0,
            max_value=700,
            value=200,
            step=1,
        )
        fasting_bs = st.selectbox(
            "Fasting Blood Sugar > 120 mg/dl",
            options=[0, 1],
            format_func=lambda x: "Yes (1)" if x == 1 else "No (0)",
        )

    with col2:
        resting_ecg = st.selectbox(
            "Resting ECG Results",
            options=["Normal", "ST", "LVH"],
            help="Normal: Normal, ST: ST-T wave abnormality, LVH: Left ventricular hypertrophy",
        )
        max_hr = st.number_input(
            "Maximum Heart Rate Achieved (MaxHR)",
            min_value=60,
            max_value=220,
            value=150,
            step=1,
        )
        exercise_angina = st.selectbox(
            "Exercise-Induced Angina",
            options=["N", "Y"],
            help="N: No, Y: Yes",
        )
        oldpeak = st.number_input(
            "Oldpeak (ST Depression)",
            min_value=-3.0,
            max_value=10.0,
            value=0.0,
            step=0.1,
        )
        st_slope = st.selectbox(
            "ST Slope",
            options=["Up", "Flat", "Down"],
            help="Slope of peak exercise ST segment",
        )

    st.markdown("---")

    # 5. Predict Button Trigger
    predict_button = st.button("Predict Heart Disease", type="primary")

    if predict_button:
        try:
            # Create raw input DataFrame
            input_df = pd.DataFrame(
                [
                    {
                        "Age": age,
                        "Sex": sex,
                        "ChestPainType": chest_pain,
                        "RestingBP": resting_bp,
                        "Cholesterol": cholesterol,
                        "FastingBS": fasting_bs,
                        "RestingECG": resting_ecg,
                        "MaxHR": max_hr,
                        "ExerciseAngina": exercise_angina,
                        "Oldpeak": oldpeak,
                        "ST_Slope": st_slope,
                    }
                ]
            )

            # Detect categorical columns and apply One-Hot Encoding
            cat_cols = input_df.select_dtypes(include=["object"]).columns.tolist()
            input_encoded = pd.get_dummies(
                input_df, columns=cat_cols, drop_first=True
            ).astype(int)

            # Reindex to match the training feature columns from 'columns.pkl'
            input_aligned = input_encoded.reindex(
                columns=encoded_columns, fill_value=0
            )

            # Make prediction
            prediction = model.predict(input_aligned)[0]
            probability = model.predict_proba(input_aligned)[0]

            st.markdown("### Diagnosis Result")
            if prediction == 1:
                st.error(
                    f"🚨 **High Risk**: Heart Disease Detected! (Confidence: {probability[1] * 100:.2f}%)"
                )
            else:
                st.success(
                    f"✅ **Low Risk**: No Heart Disease Detected. (Confidence: {probability[0] * 100:.2f}%)"
                )

        except Exception as e:
            st.error(f"An error occurred while making predictions: {e}")
```
